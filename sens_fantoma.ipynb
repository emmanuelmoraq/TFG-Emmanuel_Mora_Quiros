## 3. Espectro energético de SpekPy

Se lee el espectro exportado para la configuración final y se normaliza a probabilidades por bin. La intensidad absoluta no se toma del archivo de espectro; se incorpora posteriormente mediante $K_{a,r}$.


```python

def read_spekpy_spectrum(filename):
    filename = Path(filename)

    if not filename.exists():
        raise FileNotFoundError(
            "No se encontró el archivo de espectro.\n"
            f"Ruta buscada: {filename.resolve()}"
        )

    energies = []
    fluences = []

    with open(filename, "r", encoding="utf-8") as f:
        for line in f:
            line = line.strip()

            if not line:
                continue

            # Admite espacios o tabulaciones.
            # También admite coma decimal.
            parts = line.replace(",", ".").split()

            if len(parts) < 2:
                continue

            try:
                energy = float(parts[0])
                fluence = float(parts[1])
            except ValueError:
                # Encabezados/comentarios.
                continue

            # Se conservan también bins de fluencia cero.
            if energy > 0 and fluence >= 0:
                energies.append(energy)
                fluences.append(fluence)

    energies = np.asarray(energies, dtype=float)
    fluences = np.asarray(fluences, dtype=float)

    if energies.size == 0:
        raise ValueError(
            "No se encontraron dos columnas numéricas válidas "
            "en el archivo de SpekPy."
        )

    if np.sum(fluences) <= 0:
        raise ValueError(
            "La suma de las fluencias del espectro es cero."
        )

    # La tabla de SpekPy representa fluencia diferencial.
    # La multiplicación por dE hace el lector compatible también
    # con un espaciado energético no perfectamente uniforme.
    dE = np.gradient(energies)

    weights = fluences * dE
    weights = weights / np.sum(weights)

    return energies, weights, fluences


energies_keV, spectrum_weights, spectrum_fluences = (
    read_spekpy_spectrum(SPECTRUM_FILE)
)

mean_energy_from_file = np.sum(
    energies_keV * spectrum_weights
)

print("Espectro encontrado:", SPECTRUM_FILE.resolve())
print("Número de bins:", len(energies_keV))
print("Energía mínima:", energies_keV.min(), "keV")
print("Energía máxima:", energies_keV.max(), "keV")
print("Suma de probabilidades:", spectrum_weights.sum())
print("Energía media calculada:", mean_energy_from_file, "keV")
print("Energía media SpekPy esperada:", SPEKPY_MEAN_ENERGY_KEV, "keV")

if energies_keV.max() > TUBE_VOLTAGE_KV + 1.0:
    print(
        "ADVERTENCIA: la energía máxima supera de forma "
        "inesperada el kVp configurado."
    )


plt.figure(figsize=(8, 5))
plt.step(
    energies_keV,
    spectrum_weights,
    where="mid"
)
plt.xlabel("Energía [keV]")
plt.ylabel("Probabilidad por bin")
plt.title("Espectro del Artis Zee - 73,3 kVp")
plt.grid(alpha=0.3)
plt.tight_layout()

if SAVE_FIGURES:
    plt.savefig(
        FIGURE_DIR / "01_espectro_80kVp.png",
        dpi=FIGURE_DPI,
        bbox_inches="tight"
    )

plt.show()
```

## 4. Geometría del haz y normalización de referencia

Se calculan la orientación LAO, la posición del foco y del receptor, la apertura angular del haz y la posición del IRP. El signo de `LAO_DEG` responde al sistema de coordenadas definido en el modelo.


```python

# ============================================================
# ROTACIÓN LAO
# ============================================================

R_lao = Rotation.from_euler(
    "y",
    LAO_DEG,
    degrees=True
)

beam_direction = R_lao.apply(
    np.array([0.0, 0.0, 1.0])
)


# ============================================================
# FOCO E ISOCENTRO
# ============================================================

source_position = (
    -SOURCE_TO_ISOCENTER_CM
    * cm
    * beam_direction
)

ISO_TO_DETECTOR_CM = (
    SID_CM - SOURCE_TO_ISOCENTER_CM
)

detector_position = (
    ISO_TO_DETECTOR_CM
    * cm
    * beam_direction
)


# ============================================================
# APERTURA REAL DEL HAZ
# ============================================================
#
# Se usa el campo REAL medido de 20.74 cm a 100 cm.
# No se vuelve a sobrescribir theta_max con el FOV nominal.

theta_max = np.arctan(
    (FIELD_MEASURED_CM / 2)
    / FIELD_REFERENCE_DISTANCE_CM
) * gate.g4_units.rad

field_iso_cm = (
    2.0
    * SOURCE_TO_ISOCENTER_CM
    * np.tan(theta_max)
)


# ============================================================
# IRP Y Ka,r DE REFERENCIA
# ============================================================

IRP_FROM_ISOCENTER_TOWARD_SOURCE_CM = 15.0

IRP_SOURCE_DISTANCE_CM = (
    SOURCE_TO_ISOCENTER_CM
    - IRP_FROM_ISOCENTER_TOWARD_SOURCE_CM
)

KA_R_SPEKPY_MGY = (
    SPEKPY_KERMA_UGY_PER_MAS
    * MAS
    * (
        SPEKPY_REFERENCE_DISTANCE_CM
        / IRP_SOURCE_DISTANCE_CM
    )**2
    / 1000.0
)

if KA_R_MEASURED_MGY is None:
    KA_R_REFERENCE_MGY = KA_R_SPEKPY_MGY
    KA_R_REFERENCE_SOURCE = "SpekPy (estimación del modelo)"
else:
    KA_R_REFERENCE_MGY = float(KA_R_MEASURED_MGY)
    KA_R_REFERENCE_SOURCE = "CAREreport/RDSR o medición"


print("Dirección central del haz:", beam_direction)
print("Posición del foco [cm]:", source_position / cm)
print("Posición del detector [cm]:", detector_position / cm)
print("SID calculada [cm]:",
      np.linalg.norm(detector_position - source_position) / cm)
print("Semiángulo del haz [°]:", theta_max / deg)
print("Campo estimado en isocentro [cm]:", field_iso_cm)
print("Distancia foco-IRP [cm]:", IRP_SOURCE_DISTANCE_CM)
print("Ka,r de referencia [mGy]:", KA_R_REFERENCE_MGY)
print("Fuente de normalización:", KA_R_REFERENCE_SOURCE)
```

## 5. Fantoma antropomórfico simplificado y materiales

El torso del CIRS se representa mediante geometrías simples y materiales equivalentes para tejido blando, pulmón y hueso. Esta simplificación se mantiene explícita como una limitación del modelo.


```python

# ============================================================
# FANTOMA CIRS ATOM 702 - MODELO TORÁCICO SIMPLIFICADO
# ============================================================

THORAX_WIDTH_CM = 25.0
THORAX_LENGTH_CM = 30.0
THORAX_AP_CM = 20.0

PHANTOM_CENTER_X_CM = 0.0
PHANTOM_CENTER_Y_CM = 0.0


def add_cirs_702_materials(sim):

    g_cm3 = gate.g4_units.g_cm3
    db = sim.volume_manager.material_database

    def normalize(weights):
        weights = np.asarray(weights, dtype=float)
        return (weights / np.sum(weights)).tolist()

    # Tejido blando
    db.add_material_weights(
        "CIRS_SoftTissue",
        ["H", "C", "N", "O", "Mg", "Cl"],
        normalize([
            8.47,
            57.44,
            1.65,
            24.59,
            7.62,
            0.19
        ]),
        1.055 * g_cm3
    )

    # Pulmón
    db.add_material_weights(
        "CIRS_Lung",
        ["H", "C", "N", "O", "Cl", "Ca"],
        normalize([
            8.59,
            65.89,
            3.52,
            19.29,
            1.69,
            1.01
        ]),
        0.202 * g_cm3
    )

    # Hueso
    db.add_material_weights(
        "CIRS_Bone",
        ["H", "C", "N", "O", "Mg", "Cl", "Ca"],
        normalize([
            4.83,
            37.03,
            0.97,
            35.66,
            6.19,
            0.05,
            15.24
        ]),
        1.517 * g_cm3
    )


PHANTOM_CENTER_Z_CM = (
    TABLE_TOP_Z_CM + THORAX_AP_CM / 2
)

print(
    "Centro del torso [cm]:",
    (
        PHANTOM_CENTER_X_CM,
        PHANTOM_CENTER_Y_CM,
        PHANTOM_CENTER_Z_CM,
    )
)
print(
    "Rango vertical aproximado del torso [cm]:",
    (
        PHANTOM_CENTER_Z_CM - THORAX_AP_CM / 2,
        PHANTOM_CENTER_Z_CM + THORAX_AP_CM / 2,
    )
)
print("Plano de medición z [cm]:", DOSIMETER_Z_CM)
```

## 6. Funciones auxiliares y comprobaciones geométricas


```python

def get_point_xyz(point):
    """Devuelve [x_cm, y_cm, z_cm] para un punto experimental."""

    if isinstance(point, dict):
        x = float(point["x_cm"])
        y = float(point["y_cm"])
        z = float(
            point.get(
                "z_cm",
                DOSIMETER_Z_CM
            )
        )
        return np.array([x, y, z], dtype=float)

    point = np.asarray(point, dtype=float)

    if point.size != 3:
        raise ValueError(
            "Cada punto debe contener x, y, z."
        )

    return point



def validate_configuration():
    problems = []

    if ROOM_SIZE_X_CM <= 0:
        problems.append("ROOM_SIZE_X_CM debe ser > 0.")
    if ROOM_SIZE_Y_CM <= 0:
        problems.append("ROOM_SIZE_Y_CM debe ser > 0.")
    if ROOM_SIZE_Z_CM <= 0:
        problems.append("ROOM_SIZE_Z_CM debe ser > 0.")

    floor_z = -ISOCENTER_TO_FLOOR_CM
    ceiling_z = floor_z + ROOM_SIZE_Z_CM

    if not (floor_z <= DOSIMETER_Z_CM <= ceiling_z):
        problems.append(
            "La altura del dosímetro queda fuera de la sala."
        )

    for name, point in MEASUREMENT_POINTS.items():
        xyz = get_point_xyz(point)

        if abs(xyz[0]) > ROOM_SIZE_X_CM / 2:
            problems.append(
                f"{name}: x queda fuera del mapa."
            )

        if abs(xyz[1]) > ROOM_SIZE_Y_CM / 2:
            problems.append(
                f"{name}: y queda fuera del mapa."
            )

    if problems:
        print("ADVERTENCIAS DE CONFIGURACIÓN:")
        for p in problems:
            print(" -", p)
    else:
        print("Configuración geométrica básica: OK")

    if not MEASUREMENT_POINTS:
        print(
            "PENDIENTE: agregar los puntos experimentales "
            "en MEASUREMENT_POINTS."
        )


validate_configuration()
```

## 7. Construcción de la simulación principal

Incluye sala de aire, mesa equivalente, blindaje posterior del receptor, fantoma simplificado, fuente polienergética, `TLEDoseActor` para la malla y los puntos de medición, y `PhaseSpaceActor` en cada punto para el cálculo posterior de $H^*(10)$.


```python

def build_simulation():

    sim = gate.Simulation()

    # --------------------------------------------------------
    # CONFIGURACIÓN GENERAL
    # --------------------------------------------------------

    sim.output_dir = MAIN_OUTPUT_DIR
    sim.number_of_threads = N_THREADS

    sim.g4_verbose = False
    sim.progress_bar = True
    sim.check_volumes_overlap = True
    sim.random_seed = "auto"

    # --------------------------------------------------------
    # MUNDO
    # --------------------------------------------------------
    #
    # La sala se representa como aire. Las paredes todavía no
    # se modelan como materiales, por lo que las dimensiones
    # controlan principalmente la extensión espacial del mapa.

    floor_z_cm = -ISOCENTER_TO_FLOOR_CM
    ceiling_z_cm = floor_z_cm + ROOM_SIZE_Z_CM

    world_z_half_extent_cm = (
        max(
            abs(floor_z_cm),
            abs(ceiling_z_cm),
            abs(source_position[2] / cm)
        )
        + WORLD_MARGIN_CM
    )

    world = sim.world

    world.size = [
        (ROOM_SIZE_X_CM + 2 * WORLD_MARGIN_CM) * cm,
        (ROOM_SIZE_Y_CM + 2 * WORLD_MARGIN_CM) * cm,
        (2 * world_z_half_extent_cm) * cm,
    ]

    world.material = "G4_AIR"

    # --------------------------------------------------------
    # MATERIALES
    # --------------------------------------------------------

    add_cirs_702_materials(sim)

    # --------------------------------------------------------
    # MESA
    # --------------------------------------------------------

    table = sim.add_volume(
        "Box",
        "table"
    )

    table.size = [
        TABLE_WIDTH_CM * cm,
        TABLE_LENGTH_CM * cm,
        TABLE_EQ_AL_THICKNESS_MM * mm,
    ]

    table_center_z = (
        TABLE_TOP_Z_CM * cm
        - (TABLE_EQ_AL_THICKNESS_MM * mm) / 2
    )

    table.translation = [
        0,
        0,
        table_center_z
    ]

    table.material = "G4_Al"

    # ============================================================
    # BLINDAJE POSTERIOR DEL RECEPTOR
    # ============================================================
    
    lead_backing = sim.add_volume(
        "Box",
        "detector_lead_backing"
    )
    
    lead_backing.material = "G4_Pb"
    
    lead_backing.size = [
        LEAD_BACKING_WIDTH_CM * cm,
        LEAD_BACKING_HEIGHT_CM * cm,
        LEAD_BACKING_THICKNESS_MM * mm,
    ]
    
    # La placa debe ser perpendicular al eje del haz
    lead_backing.rotation = R_lao.as_matrix()
    
    # Centro de la placa a 30 cm del isocentro
    lead_backing.translation = (
        LEAD_BACKING_DISTANCE_ISO_CM
        * cm
        * beam_direction
    )
        

    # --------------------------------------------------------
    # FANTOMA CIRS ATOM 702 SIMPLIFICADO
    # --------------------------------------------------------

    phantom_center_x = PHANTOM_CENTER_X_CM * cm
    phantom_center_y = PHANTOM_CENTER_Y_CM * cm
    phantom_center_z = PHANTOM_CENTER_Z_CM * cm

    if PHANTOM_MODEL not in ("homogeneous", "heterogeneous"):
        raise ValueError(
            f"Modelo de fantoma no válido: {PHANTOM_MODEL}"
        )

    # ========================================================
    # TORSO - COMÚN A AMBOS MODELOS
    # ========================================================

    torso = sim.add_volume(
        "Ellipsoid",
        "phantom_torso"
    )

    torso.xSemiAxis = THORAX_WIDTH_CM / 2 * cm
    torso.ySemiAxis = THORAX_LENGTH_CM / 2 * cm
    torso.zSemiAxis = THORAX_AP_CM / 2 * cm

    torso.translation = [
        phantom_center_x,
        phantom_center_y,
        phantom_center_z
    ]

    torso.material = "CIRS_SoftTissue"

    # ========================================================
    # ESTRUCTURAS INTERNAS - MODELO HETEROGÉNEO
    # ========================================================

    if PHANTOM_MODEL == "heterogeneous":

        # ----------------------------------------------------
        # PULMÓN IZQUIERDO
        # ----------------------------------------------------

        lung_left = sim.add_volume(
            "Ellipsoid",
            "lung_left"
        )

        lung_left.mother = torso.name

        lung_left.xSemiAxis = 3.8 * cm
        lung_left.ySemiAxis = 8.5 * cm
        lung_left.zSemiAxis = 5.5 * cm

        lung_left.translation = [
            +4.5 * cm,
            0.0 * cm,
            +0.8 * cm
        ]

        lung_left.material = "CIRS_Lung"

        # ----------------------------------------------------
        # PULMÓN DERECHO
        # ----------------------------------------------------

        lung_right = sim.add_volume(
            "Ellipsoid",
            "lung_right"
        )

        lung_right.mother = torso.name

        lung_right.xSemiAxis = 3.8 * cm
        lung_right.ySemiAxis = 8.5 * cm
        lung_right.zSemiAxis = 5.5 * cm

        lung_right.translation = [
            -4.5 * cm,
            0.0 * cm,
            +0.8 * cm
        ]

        lung_right.material = "CIRS_Lung"

        # ----------------------------------------------------
        # COLUMNA VERTEBRAL
        # ----------------------------------------------------

        spine = sim.add_volume(
            "Tubs",
            "spine"
        )

        spine.mother = torso.name

        spine.rmin = 0.0 * cm
        spine.rmax = 1.3 * cm
        spine.dz = 10.0 * cm

        spine.rotation = Rotation.from_euler(
            "x",
            90,
            degrees=True
        ).as_matrix()

        spine.translation = [
            0.0 * cm,
            0.0 * cm,
            -5.8 * cm
        ]

        spine.material = "CIRS_Bone"

        # ----------------------------------------------------
        # ESTERNÓN
        # ----------------------------------------------------

        sternum = sim.add_volume(
            "Box",
            "sternum"
        )

        sternum.mother = torso.name

        sternum.size = [
            2.0 * cm,
            15.0 * cm,
            1.0 * cm
        ]

        sternum.translation = [
            0.0 * cm,
            0.0 * cm,
            +7.2 * cm
        ]

        sternum.material = "CIRS_Bone"


    # --------------------------------------------------------
    # SOPORTE LOCAL DE LA FUENTE
    # --------------------------------------------------------

    source_frame = sim.add_volume(
        "Box",
        "source_frame"
    )

    source_frame.size = [
        1 * mm,
        1 * mm,
        1 * mm
    ]

    source_frame.material = "G4_AIR"
    source_frame.translation = source_position
    source_frame.rotation = R_lao.as_matrix()

    # --------------------------------------------------------
    # FUENTE DE RAYOS X
    # --------------------------------------------------------

    source = sim.add_source(
        "GenericSource",
        "artis_zee_source"
    )

    source.attached_to = source_frame.name
    source.particle = "gamma"

    source.position.type = "point"
    source.position.translation = [0, 0, 0]

    source.direction.type = "iso"
    
    source.direction.theta = [
        180 * deg - theta_max,
        180 * deg
    ]
    
    source.direction.phi = [
        0 * deg,
        360 * deg
    ]
    
    source.direction_relative_to_attached_volume = True

    source.energy.type = "spectrum_discrete"
    source.energy.spectrum_energies = (
        energies_keV * keV
    ).tolist()
    source.energy.spectrum_weights = (
        spectrum_weights
    ).tolist()

    # La versión instalada reconoce source.n.
    source.n = int(N_PRIMARIES_PER_THREAD)

    # --------------------------------------------------------
    # FÍSICA
    # --------------------------------------------------------

    sim.physics_manager.physics_list_name = (
        "G4EmLivermorePhysics"
    )

    sim.physics_manager.set_production_cut(
        "world",
        "gamma",
        1 * mm
    )

    sim.physics_manager.set_production_cut(
        "world",
        "electron",
        1 * mm
    )

    # El torso existe en ambos modelos
    phantom_volume_names = [
        torso.name
    ]

    # Las estructuras internas solo existen en el
    # modelo heterogéneo
    if PHANTOM_MODEL == "heterogeneous":
        phantom_volume_names.extend([
            lung_left.name,
            lung_right.name,
            spine.name,
            sternum.name,
        ])

    # Aplicar cortes de producción
    for volume_name in phantom_volume_names:

        sim.physics_manager.set_production_cut(
            volume_name,
            "gamma",
            0.1 * mm
        )

        sim.physics_manager.set_production_cut(
            volume_name,
            "electron",
            0.1 * mm
        )

    sim.physics_manager.set_production_cut(
        table.name,
        "gamma",
        0.1 * mm
    )

    sim.physics_manager.set_production_cut(
        table.name,
        "electron",
        0.1 * mm
    )

    # --------------------------------------------------------
    # LIMITACIÓN DEL STEP EN AIRE
    # --------------------------------------------------------
    
    sim.physics_manager.user_limits_particles = [
        "gamma"
    ]
    
    sim.physics_manager.set_max_step_size(
        "world",
        MAX_STEP_AIR_CM * cm
    )
        
    # --------------------------------------------------------
    # MAPA DE DOSIS EN AIRE
    # --------------------------------------------------------

    nx = int(
        np.ceil(
            ROOM_SIZE_X_CM / ROOM_VOXEL_CM
        )
    )

    ny = int(
        np.ceil(
            ROOM_SIZE_Y_CM / ROOM_VOXEL_CM
        )
    )

    room_dx_cm = ROOM_SIZE_X_CM / nx
    room_dy_cm = ROOM_SIZE_Y_CM / ny

    room_dose = sim.add_actor(
        "TLEDoseActor",
        "room_dose"
    )

    room_dose.attached_to = "world"

    room_dose.size = [
        nx,
        ny,
        1
    ]

    room_dose.spacing = [
        room_dx_cm * cm,
        room_dy_cm * cm,
        ROOM_PLANE_THICKNESS_CM * cm
    ]

    room_dose.translation = [
        0,
        0,
        DOSIMETER_Z_CM * cm
    ]

    room_dose.database = "EPDL"

    # Solo necesitamos dosis e incertidumbre.
    room_dose.edep.active = False
    room_dose.dose.active = True
    room_dose.dose_uncertainty.active = True
    room_dose.dose_squared.active = True

    room_dose.dose.output_filename = (
        "room_dose_xy.mhd"
    )

    room_dose.dose_uncertainty.output_filename = (
        "room_dose_xy_uncertainty.mhd"
    )

    room_dose.counts.active = True
    room_dose.counts.output_filename = "room_counts.mhd"
    
    # --------------------------------------------------------
    # PUNTOS EXPERIMENTALES
    # --------------------------------------------------------
    
    point_actors = {}
    point_volumes = {}
    hstar_phsp_actors = {}
    

    
    for point_name, point_info in MEASUREMENT_POINTS.items():
    
        xyz_cm = get_point_xyz(
            point_info
        )
    
        # ========================================================
        # VOLUMEN FÍSICO DE AIRE
        # ========================================================
    
        point_volume = sim.add_volume(
            "Box",
            f"detector_{point_name}"
        )
    
        point_volume.size = [
            POINT_SCORING_SIZE_CM * cm,
            POINT_SCORING_SIZE_CM * cm,
            POINT_SCORING_SIZE_CM * cm
        ]
    
        point_volume.translation = [
            xyz_cm[0] * cm,
            xyz_cm[1] * cm,
            xyz_cm[2] * cm
        ]
    
        point_volume.material = "G4_AIR"
    
        # ========================================================
        # TLE DOSE ACTOR
        # ========================================================
    
        point_actor = sim.add_actor(
            "TLEDoseActor",
            f"dose_{point_name}"
        )
    
        point_actor.attached_to = (
            point_volume.name
        )
    
        point_actor.size = [
            1,
            1,
            1
        ]
    
        point_actor.spacing = [
            POINT_SCORING_SIZE_CM * cm,
            POINT_SCORING_SIZE_CM * cm,
            POINT_SCORING_SIZE_CM * cm
        ]
    
        point_actor.translation = [
            0,
            0,
            0
        ]
    
        point_actor.database = "EPDL"
    
        point_actor.dose.active = True
        point_actor.dose_squared.active = True
        point_actor.dose_uncertainty.active = True
        point_actor.edep.active = True
    
        point_actor.dose.output_filename = (
            f"{point_name}_dose.mhd"
        )
    
        point_actor.edep.output_filename = (
            f"{point_name}_edep.mhd"
        )
    
        point_actor.dose_uncertainty.output_filename = (
            f"{point_name}_uncertainty.mhd"
        )
    
        # ========================================================
        # PHASE SPACE PARA H*(10)
        # ========================================================
    
        phsp_actor = sim.add_actor(
            "PhaseSpaceActor",
            f"phsp_hstar_{point_name}"
        )
        
        phsp_actor.attached_to = (
            point_volume.name
        )
        
        phsp_actor.steps_to_store = "all"
        
        phsp_actor.attributes = [
            "EventID",
            "ParticleName",
            "PreKineticEnergy",
            "StepLength",
        ]
        
        phsp_actor.output_filename = (
            f"{point_name}_hstar_phase_space.root"
        )
    
    
        # ========================================================
        # GUARDAR ACTORES
        # ========================================================
    
        point_actors[
            point_name
        ] = point_actor
    
        point_volumes[
            point_name
        ] = point_volume
    
        hstar_phsp_actors[
            point_name
        ] = phsp_actor
    
    # --------------------------------------------------------
    # ESTADÍSTICAS
    # --------------------------------------------------------

    stats = sim.add_actor(
        "SimulationStatisticsActor",
        "Stats"
    )

    stats.track_types_flag = True
    stats.output_filename = "statistics.txt"

    
    
    return (
        sim,
        source,
        torso,
        table,
        room_dose,
        point_actors,
        hstar_phsp_actors,
        stats
    )
```

## 8. Simulación independiente de normalización en el IRP

Esta simulación calcula la respuesta por primario en el IRP mediante `TLEDoseActor` y registra el cruce de fotones mediante `PhaseSpaceActor`. La geometría del IRP se mantiene separada de los detectores ocupacionales.


```python
def build_air_kerma_simulation():

    sim = gate.Simulation()

    sim.output_dir = NORM_OUTPUT_DIR
    sim.number_of_threads = N_THREADS

    sim.g4_verbose = False
    sim.progress_bar = True
    sim.check_volumes_overlap = True
    sim.random_seed = "auto"

    # --------------------------------------------------------
    # MUNDO
    # --------------------------------------------------------

    world = sim.world

    world.size = [
        3 * m,
        3 * m,
        3 * m
    ]

    world.material = "G4_AIR"

    # --------------------------------------------------------
    # SOPORTE LOCAL DE LA FUENTE
    # --------------------------------------------------------

    source_frame = sim.add_volume(
        "Box",
        "source_frame_norm"
    )

    source_frame.size = [
        1 * mm,
        1 * mm,
        1 * mm
    ]

    source_frame.material = "G4_AIR"

    source_frame.translation = (
        source_position
    )

    source_frame.rotation = (
        R_lao.as_matrix()
    )

    # --------------------------------------------------------
    # FUENTE
    # --------------------------------------------------------

    source = sim.add_source(
        "GenericSource",
        "artis_zee_source_norm"
    )

    source.attached_to = (
        source_frame.name
    )

    source.particle = "gamma"

    source.position.type = "point"

    source.position.translation = [
        0,
        0,
        0
    ]

    # Haz cónico
    source.direction.type = "iso"
    
    source.direction.theta = [
        180 * deg - theta_max,
        180 * deg
    ]
    
    source.direction.phi = [
        0 * deg,
        360 * deg
    ]
    
    source.direction_relative_to_attached_volume = True

    # --------------------------------------------------------
    # ESPECTRO
    # --------------------------------------------------------

    source.energy.type = (
        "spectrum_discrete"
    )

    source.energy.spectrum_energies = (
        energies_keV * keV
    ).tolist()

    source.energy.spectrum_weights = (
        spectrum_weights
    ).tolist()

    # --------------------------------------------------------
    # HISTORIAS
    # --------------------------------------------------------

    source.n = int(
        N_PRIMARIES_PER_THREAD_NORMALIZATION
    )

    # --------------------------------------------------------
    # FÍSICA
    # --------------------------------------------------------

    sim.physics_manager.physics_list_name = (
        "G4EmLivermorePhysics"
    )

    sim.physics_manager.set_production_cut(
        "world",
        "gamma",
        1 * mm
    )

    sim.physics_manager.set_production_cut(
        "world",
        "electron",
        1 * mm
    )

    # --------------------------------------------------------
    # LIMITACIÓN DEL STEP EN EL AIRE
    # --------------------------------------------------------

    sim.physics_manager.user_limits_particles = [
        "gamma"
    ]

    sim.physics_manager.set_max_step_size(
        "world",
        MAX_STEP_AIR_CM * cm
    )

    # --------------------------------------------------------
    # POSICIÓN DEL IRP
    # --------------------------------------------------------

    irp_position = (
        -IRP_FROM_ISOCENTER_TOWARD_SOURCE_CM
        * cm
        * beam_direction
    )

    print(
        "IRP [cm] =",
        irp_position / cm
    )

    print(
        "Distancia foco-IRP [cm] =",
        np.linalg.norm(
            irp_position - source_position
        ) / cm
    )

    # --------------------------------------------------------
    # VOLUMEN FÍSICO DE AIRE EN EL IRP
    # --------------------------------------------------------

    irp_detector = sim.add_volume(
        "Box",
        "irp_air_detector"
    )

    irp_detector.size = [
        IRP_DETECTOR_WIDTH_CM * cm,
        IRP_DETECTOR_WIDTH_CM * cm,
        IRP_DETECTOR_THICKNESS_CM * cm
    ]
    
    irp_detector.translation = (
        irp_position.tolist()
    )
    
    irp_detector.rotation = (
        R_lao.as_matrix()
    )
    irp_detector.material = "G4_AIR"

    irp_detector.color = [
        1.0,
        0.0,
        0.0,
        0.7
    ]

    # ============================================================
    # TLE DOSE ACTOR
    # ============================================================

    irp_dose = sim.add_actor(
        "TLEDoseActor",
        "irp_air_dose"
    )

    irp_dose.attached_to = (
        irp_detector.name
    )

    irp_dose.size = [
        1,
        1,
        1
    ]

    irp_dose.spacing = [
        IRP_DETECTOR_WIDTH_CM * cm,
        IRP_DETECTOR_WIDTH_CM * cm,
        IRP_DETECTOR_THICKNESS_CM * cm
    ]

    irp_dose.translation = [
        0,
        0,
        0
    ]

    irp_dose.database = "EPDL"

    # --------------------------------------------------------
    # SALIDAS TLE
    # --------------------------------------------------------

    irp_dose.dose.active = True
    irp_dose.dose_squared.active = True
    irp_dose.dose_uncertainty.active = True
    irp_dose.edep.active = True

    irp_dose.dose.output_filename = (
        "irp_air_dose.mhd"
    )

    irp_dose.edep.output_filename = (
        "irp_air_edep.mhd"
    )

    irp_dose.dose_uncertainty.output_filename = (
        "irp_air_dose_uncertainty.mhd"
    )

    # ============================================================
    # PHASE SPACE ACTOR DIRECTAMENTE EN EL IRP
    # ============================================================

    phsp_irp = sim.add_actor(
        "PhaseSpaceActor",
        "phsp_irp"
    )

    phsp_irp.attached_to = (
        irp_detector.name
    )

    # Guardar únicamente partículas al entrar
    # al volumen del IRP.
    phsp_irp.steps_to_store = "entering"

    phsp_irp.attributes = [
        "EventID",
        "TrackID",
        "ParticleName",
        "KineticEnergy",
        "PrePosition",
        "PreDirection",
    ]

    phsp_irp.output_filename = (
        "irp_phase_space.root"
    )


    # --------------------------------------------------------
    # ESTADÍSTICAS
    # --------------------------------------------------------

    stats = sim.add_actor(
        "SimulationStatisticsActor",
        "Stats_Normalization"
    )

    stats.track_types_flag = True

    stats.output_filename = (
        "statistics_normalization.txt"
    )

    # --------------------------------------------------------
    # RETORNO
    # --------------------------------------------------------

    return (
        sim,
        source,
        irp_detector,
        irp_dose,
        phsp_irp,
        stats
    )
```

```python


def read_mhd_array(filename):

    filename = Path(filename)

    if not filename.exists():
        raise FileNotFoundError(
            f"No se encontró el archivo: {filename}"
        )

    image = sitk.ReadImage(
        str(filename)
    )

    array = sitk.GetArrayFromImage(
        image
    )

    return array



# ============================================================
# SIMULACIÓN PRINCIPAL
# ============================================================

(
    sim,
    source,
    torso,
    table,
    room_dose,
    point_actors,
    hstar_phsp_actors,
    stats
) = build_simulation()


print("Árbol de volúmenes:")

try:
    print(
        sim.volume_manager.dump_volume_tree()
    )
except Exception:
    pass


# Ejecutar en proceso independiente
sim.run(
    start_new_process=True
)


print(
    "\nESTADÍSTICAS SIMULACIÓN PRINCIPAL"
)

print(stats)


# ============================================================
# SIMULACIÓN DE NORMALIZACIÓN
# ============================================================

(
    sim_norm,
    source_norm,
    irp_detector,
    irp_dose,
    phsp_irp,
    stats_norm
) = build_air_kerma_simulation()


# Ejecutar también en proceso independiente
sim_norm.run(
    start_new_process=True
)

print(
    "\nESTADÍSTICAS NORMALIZACIÓN"
)

print(stats_norm)


# ============================================================
# LECTURA DE RESULTADOS DEL IRP
# ============================================================

irp_dose_file = (
    NORM_OUTPUT_DIR
    / "irp_air_dose.mhd"
)

irp_edep_file = (
    NORM_OUTPUT_DIR
    / "irp_air_edep.mhd"
)

irp_uncertainty_file = (
    NORM_OUTPUT_DIR
    / "irp_air_dose_uncertainty.mhd"
)


# Leer los archivos MHD
irp_dose_array = read_mhd_array(
    irp_dose_file
)

irp_edep_array = read_mhd_array(
    irp_edep_file
)

irp_uncertainty_array = read_mhd_array(
    irp_uncertainty_file
)


# ============================================================
# EXTRAER EL ÚNICO VOXEL
# ============================================================

irp_dose_raw = float(
    np.squeeze(
        irp_dose_array
    )
)

irp_edep_raw = float(
    np.squeeze(
        irp_edep_array
    )
)

irp_uncertainty = float(
    np.squeeze(
        irp_uncertainty_array
    )
)


# ============================================================
# MOSTRAR RESULTADOS
# ============================================================

print("\nRESULTADOS DE NORMALIZACIÓN")

print(
    f"Dosis IRP MC [Gy] = "
    f"{irp_dose_raw:.6e}"
)

print(
    f"Edep IRP MC = "
    f"{irp_edep_raw:.6e}"
)

print(
    f"Incertidumbre relativa IRP = "
    f"{irp_uncertainty:.6e}"
)

print(
    f"Incertidumbre relativa IRP [%] = "
    f"{irp_uncertainty * 100:.2f} %"
)


# ============================================================
# COMPROBACIÓN
# ============================================================

if irp_dose_raw <= 0:

    if irp_edep_raw > 0:

        raise RuntimeError(
            "Se obtuvo energía depositada en el detector "
            "del IRP, pero la dosis es cero. "
            "Se debe revisar la configuración del TLEDoseActor."
        )

    else:

        raise RuntimeError(
            "La dosis y la energía depositada en el IRP "
            "son cero. El haz no está atravesando "
            "correctamente el detector de normalización."
        )


print(
    "\nNormalización IRP calculada correctamente."
)
```

## 11. Lectura de salidas y normalización absoluta

Se obtiene la respuesta del IRP por primario y el número equivalente de primarios asociado con el $K_{a,r}$ de referencia. Este factor se utiliza para escalar los resultados de la simulación principal.


```python

def actor_image_to_array(output_interface):
    """Convierte la salida de imagen de un actor a NumPy."""

    # GATE 10 permite acceder a la imagen directamente
    # después de sim.run().
    try:
        image = output_interface.image

        if image is not None:
            return np.asarray(image)

    except Exception:
        pass

    # Alternativa: leer el archivo escrito en disco.
    try:
        import SimpleITK as sitk

        output_path = (
            output_interface.get_output_path()
        )

        image = sitk.ReadImage(
            str(output_path)
        )

        return sitk.GetArrayFromImage(
            image
        )

    except Exception as exc:
        raise RuntimeError(
            "No fue posible obtener la imagen del actor "
            "ni desde memoria ni desde disco."
        ) from exc


# ============================================================
# MAPA PRINCIPAL
# ============================================================

room_dose_raw = np.squeeze(
    actor_image_to_array(
        room_dose.dose
    )
)

room_uncertainty = np.squeeze(
    actor_image_to_array(
        room_dose.dose_uncertainty
    )
)


# ============================================================
# NORMALIZACIÓN EN EL IRP
# ============================================================

irp_dose_raw = float(
    np.squeeze(
        actor_image_to_array(
            irp_dose.dose
        )
    )
)

irp_uncertainty = float(
    np.squeeze(
        actor_image_to_array(
            irp_dose.dose_uncertainty
        )
    )
)

if irp_dose_raw <= 0:
    raise RuntimeError(
        "El actor de normalización obtuvo dosis cero "
        "en el IRP. Se debe revisar su posición/tamaño."
    )

# Resultado MC por primario:
irp_dose_per_primary_Gy = (
    irp_dose_raw
    / N_PRIMARIES_NORMALIZATION
)

room_dose_per_primary_Gy = (
    room_dose_raw
    / N_PRIMARIES
)

# Ka,r de referencia para la exposición real.
ka_r_reference_Gy = (
    KA_R_REFERENCE_MGY * 1e-3
)

# Número equivalente de primarios reales que producirían
# el Ka,r de referencia.
equivalent_real_primaries = (
    ka_r_reference_Gy
    / irp_dose_per_primary_Gy
)

# Dosis absorbida en aire / aproximación de kerma colisional
# para la exposición experimental completa.
room_dose_absolute_Gy = (
    room_dose_per_primary_Gy
    * equivalent_real_primaries
)

room_dose_absolute_uGy = (
    room_dose_absolute_Gy
    * 1e6
)

# Tasa equivalente de dosis absorbida en aire.
room_dose_rate_mGy_h = (
    room_dose_absolute_Gy
    * 1e3
    * 3600.0
    / EXPOSURE_TIME_S
)

room_uncertainty_percent = (
    room_uncertainty * 100.0
)


print("Ka,r usado [mGy]:", KA_R_REFERENCE_MGY)
print("Fuente:", KA_R_REFERENCE_SOURCE)
print("IRP MC por primario [Gy]:", irp_dose_per_primary_Gy)
print(
    "Incertidumbre relativa del IRP [%]:",
    irp_uncertainty * 100.0
)
print(
    "Máximo del mapa [uGy/exposición]:",
    np.nanmax(room_dose_absolute_uGy)
)
```

## 12. Cálculo de $\dot H^*(10)$ y validación puntual

La fluencia se estima mediante longitud de trayectoria dentro de cada volumen de puntuación. Las contribuciones se ponderan con coeficientes de conversión fluencia--$H^*(10)$ y se agrupan por historia Monte Carlo para estimar la incertidumbre estadística. La comparación principal se realiza contra los valores experimentales en las mismas posiciones.


```python


# ============================================================
# COEFICIENTES ICRP 74:
# H*(10) / fluencia para fotones
#
# Energía: MeV
# Coeficiente: pSv cm²
# ============================================================

HSTAR_ENERGY_MEV = np.array([
    0.010,
    0.015,
    0.020,
    0.030,
    0.040,
    0.050,
    0.060,
    0.080,
    0.100,
    0.150,
    0.200,
    0.300,
    0.400,
    0.500,
    0.600,
    0.800,
    1.000,
])

HSTAR_COEFF_PSV_CM2 = np.array([
    0.061,
    0.83,
    1.05,
    0.81,
    0.64,
    0.55,
    0.51,
    0.53,
    0.61,
    0.89,
    1.20,
    1.80,
    2.38,
    2.93,
    3.44,
    4.38,
    5.20,
])


# ============================================================
# INTERPOLACIÓN LOG-LOG
# ============================================================

def hstar_coeff_from_energy_mev(energy_mev):

    energy_mev = np.asarray(
        energy_mev,
        dtype=float
    )

    coeff = np.full(
        energy_mev.shape,
        np.nan,
        dtype=float
    )

    valid = (
        (energy_mev >= HSTAR_ENERGY_MEV.min())
        &
        (energy_mev <= HSTAR_ENERGY_MEV.max())
    )

    coeff[valid] = np.exp(
        np.interp(
            np.log(
                energy_mev[valid]
            ),
            np.log(
                HSTAR_ENERGY_MEV
            ),
            np.log(
                HSTAR_COEFF_PSV_CM2
            )
        )
    )

    return coeff


# ============================================================
# FUNCIÓN PARA UN PUNTO
# ============================================================

def calculate_hstar10_point(
    root_path,
    n_primaries,
    equivalent_real_primaries,
    exposure_time_s,
    scoring_size_cm,
):

    volume_cm3 = (
        scoring_size_cm ** 3
    )

    # --------------------------------------------------------
    # LEER PHASE SPACE
    # --------------------------------------------------------

    with uproot.open(root_path) as f:

        tree = f[f.keys()[0]]

        arrays = tree.arrays(
            [
                "EventID",
                "ParticleName",
                "PreKineticEnergy",
                "StepLength",
            ],
            library="np"
        )

    event_id = arrays["EventID"]
    particle = arrays["ParticleName"]
    energy_mev = arrays["PreKineticEnergy"]
    step_length_mm = arrays["StepLength"]

    # --------------------------------------------------------
    # NORMALIZAR ParticleName A STRING
    # --------------------------------------------------------

    particle_str = np.array([
        (
            p.decode()
            if isinstance(p, bytes)
            else str(p)
        )
        for p in particle
    ])

    # --------------------------------------------------------
    # SOLO GAMMAS
    # --------------------------------------------------------

    gamma_mask = (
        particle_str == "gamma"
    )

    event_id = event_id[
        gamma_mask
    ]

    energy_mev = energy_mev[
        gamma_mask
    ]

    step_length_mm = step_length_mm[
        gamma_mask
    ]

    # --------------------------------------------------------
    # CONVERSIÓN mm -> cm
    # --------------------------------------------------------

    step_length_cm = (
        step_length_mm
        / 10.0
    )

    # --------------------------------------------------------
    # COEFICIENTES H*(10)
    # --------------------------------------------------------

    h_coeff = (
        hstar_coeff_from_energy_mev(
            energy_mev
        )
    )

    # --------------------------------------------------------
    # EVENTOS FUERA DE RANGO
    # --------------------------------------------------------

    valid = (
        np.isfinite(h_coeff)
        &
        np.isfinite(step_length_cm)
        &
        (step_length_cm > 0)
    )

    n_gamma_total = len(
        energy_mev
    )

    n_gamma_valid = np.sum(
        valid
    )

    fraction_valid = (
        n_gamma_valid
        / n_gamma_total
        if n_gamma_total > 0
        else np.nan
    )

    event_id = event_id[
        valid
    ]

    step_length_cm = (
        step_length_cm[
            valid
        ]
    )

    h_coeff = h_coeff[
        valid
    ]

    energy_valid_mev = (
        energy_mev[
            valid
        ]
    )

    # ========================================================
    # ESTIMADOR DE FLUENCIA POR LONGITUD DE TRAYECTORIA
    #
    # Phi = sum(L) / V
    #
    # Contribución:
    #
    # (cm / cm³) * (pSv cm²) = pSv
    # ========================================================

    hstar_step_pSv = (
        step_length_cm
        / volume_cm3
        * h_coeff
    )

    # ========================================================
    # SUMA POR HISTORIA MONTE CARLO
    # ========================================================

    unique_events, inverse = np.unique(
        event_id,
        return_inverse=True
    )

    hstar_per_event_pSv = np.bincount(
        inverse,
        weights=hstar_step_pSv
    )

    # --------------------------------------------------------
    # SUMA TOTAL DE LA CORRIDA
    # --------------------------------------------------------

    S1 = np.sum(
        hstar_per_event_pSv
    )

    S2 = np.sum(
        hstar_per_event_pSv ** 2
    )

    N = float(
        n_primaries
    )

    # ========================================================
    # H*(10) POR PRIMARIO
    # ========================================================

    hstar_per_primary_pSv = (
        S1 / N
    )

    # ========================================================
    # INCERTIDUMBRE ESTADÍSTICA
    #
    # Se incluyen implícitamente las historias con contribución 0.
    # ========================================================

    if N > 1 and S1 > 0:

        sample_variance = (
            S2
            - (S1 ** 2) / N
        ) / (
            N - 1
        )

        sample_variance = max(
            sample_variance,
            0.0
        )

        standard_error_mean = np.sqrt(
            sample_variance
            / N
        )

        relative_uncertainty = (
            standard_error_mean
            / hstar_per_primary_pSv
        )

    else:

        relative_uncertainty = np.nan

    # ========================================================
    # ESCALADO A LA EXPOSICIÓN REAL
    # ========================================================

    hstar_real_pSv = (
        hstar_per_primary_pSv
        * equivalent_real_primaries
    )

    # pSv -> mSv
    hstar_real_mSv = (
        hstar_real_pSv
        * 1e-9
    )

    # ========================================================
    # TASA H*(10)
    # ========================================================

    hstar_rate_mSv_h = (
        hstar_real_mSv
        * 3600.0
        / exposure_time_s
    )

    return {

        "Hstar_per_primary_pSv":
            hstar_per_primary_pSv,

        "Hstar_real_mSv":
            hstar_real_mSv,

        "Hstar_rate_mSv_h":
            hstar_rate_mSv_h,

        "u_MC_percent":
            relative_uncertainty
            * 100.0,

        "N_gamma_total":
            n_gamma_total,

        "N_gamma_valid":
            n_gamma_valid,

        "fraction_valid":
            fraction_valid,

        "energy_min_keV":
            (
                np.min(
                    energy_valid_mev
                )
                * 1000.0
                if len(
                    energy_valid_mev
                ) > 0
                else np.nan
            ),

        "energy_max_keV":
            (
                np.max(
                    energy_valid_mev
                )
                * 1000.0
                if len(
                    energy_valid_mev
                ) > 0
                else np.nan
            ),

        "N_events_nonzero":
            len(
                unique_events
            ),
    }


# ============================================================
# CALCULAR TODOS LOS PUNTOS
# ============================================================

def safe_float(value):

    try:
        return float(value)

    except (TypeError, ValueError):

        return np.nan

rows = []

for point_name, point_info in (
    MEASUREMENT_POINTS.items()
):

    root_path = (
        Path(MAIN_OUTPUT_DIR)
        / f"{point_name}_hstar_phase_space.root"
    )

    if not root_path.exists():

        print(
            f"ADVERTENCIA: no existe {root_path}"
        )

        continue

    result = (
        calculate_hstar10_point(
            root_path=root_path,
            n_primaries=N_PRIMARIES,
            equivalent_real_primaries=
                equivalent_real_primaries,
            exposure_time_s=
                EXPOSURE_TIME_S,
            scoring_size_cm=
                POINT_SCORING_SIZE_CM,
        )
    )

    # --------------------------------------------------------
    # DATOS EXPERIMENTALES
    # --------------------------------------------------------
    
    measured = safe_float(
        point_info.get(
            "measured_msv_h",
            np.nan
        )
    )
    
    u_exp = safe_float(
        point_info.get(
            "u_measured_msv_h",
            np.nan
        )
    )
    
    mc = result[
        "Hstar_rate_mSv_h"
    ]
    
    # --------------------------------------------------------
    # DIFERENCIA RELATIVA
    # --------------------------------------------------------
    
    if (
        np.isfinite(measured)
        and measured > 0
    ):
    
        difference_percent = (
            100.0
            * (
                mc - measured
            )
            / measured
        )
    
    else:
    
        difference_percent = np.nan
    
    # --------------------------------------------------------
    # INCERTIDUMBRE MC ABSOLUTA
    # --------------------------------------------------------
    
    u_mc_abs = (
        mc
        * result[
            "u_MC_percent"
        ]
        / 100.0
    )
    
    # El error normalizado no se calcula aquí porque las incertidumbres
    # MC y experimentales no se han armonizado necesariamente al mismo
    # nivel de cobertura.
    En = np.nan

    rows.append(
        {
            "Punto":
                point_name,

            "MC_Hstar_mSv_h":
                mc,

            "u_MC_percent":
                result[
                    "u_MC_percent"
                ],

            "Experimental_Hstar_mSv_h":
                measured,

            "u_exp_mSv_h":
                u_exp,

            "Diferencia_percent":
                difference_percent,

            "Hstar_pSv_per_primary":
                result[
                    "Hstar_per_primary_pSv"
                ],

            "N_eventos_contribuyen":
                result[
                    "N_events_nonzero"
                ],

            "N_steps_gamma":
                result[
                    "N_gamma_valid"
                ],

            "Fraccion_energia_valida":
                result[
                    "fraction_valid"
                ],

            "Emin_keV":
                result[
                    "energy_min_keV"
                ],

            "Emax_keV":
                result[
                    "energy_max_keV"
                ],
        }
    )


# ============================================================
# TABLA
# ============================================================

df_hstar = pd.DataFrame(
    rows
)

print(
    "\n========== H*(10) MONTE CARLO VS EXPERIMENTO ==========\n"
)

display(
    df_hstar[
        [
            "Punto",
            "MC_Hstar_mSv_h",
            "u_MC_percent",
            "Experimental_Hstar_mSv_h",
            "u_exp_mSv_h",
            "Diferencia_percent",
        ]
    ]
)


# ============================================================
# COMPARACIÓN RELATIVA
# ============================================================

if len(df_hstar) > 0:

    df_hstar[
        "MC_rel_percent"
    ] = (
        100.0
        * df_hstar[
            "MC_Hstar_mSv_h"
        ]
        / df_hstar[
            "MC_Hstar_mSv_h"
        ].max()
    )

    df_hstar[
        "EXP_rel_percent"
    ] = (
        100.0
        * df_hstar[
            "Experimental_Hstar_mSv_h"
        ]
        / df_hstar[
            "Experimental_Hstar_mSv_h"
        ].max()
    )

    df_hstar[
        "Diferencia_relativa_patron"
    ] = (
        df_hstar[
            "MC_rel_percent"
        ]
        -
        df_hstar[
            "EXP_rel_percent"
        ]
    )

    print(
        "\n========== COMPARACIÓN ESPACIAL RELATIVA ==========\n"
    )

    display(
        df_hstar[
            [
                "Punto",
                "MC_rel_percent",
                "EXP_rel_percent",
                "Diferencia_relativa_patron",
            ]
        ]
    )


# ============================================================
# CONTROL DE CALIDAD DEL PHASE SPACE
# ============================================================

print(
    "\n========== CONTROL DE CALIDAD H*(10) ==========\n"
)

display(
    df_hstar[
        [
            "Punto",
            "N_eventos_contribuyen",
            "N_steps_gamma",
            "Fraccion_energia_valida",
            "Emin_keV",
            "Emax_keV",
        ]
    ]
)


# ============================================================
# RESUMEN
# ============================================================

if len(df_hstar) > 0:

    valid_diff = (
        df_hstar[
            "Diferencia_percent"
        ].dropna()
    )

    valid_unc = (
        df_hstar[
            "u_MC_percent"
        ].dropna()
    )

    print(
        "\n========== RESUMEN =========="
    )

    print(
        "Diferencia absoluta media [%] =",
        np.mean(
            np.abs(
                valid_diff
            )
        )
        if len(valid_diff) > 0
        else np.nan
    )

    print(
        "Diferencia absoluta máxima [%] =",
        np.max(
            np.abs(
                valid_diff
            )
        )
        if len(valid_diff) > 0
        else np.nan
    )

    print(
        "Incertidumbre MC media [%] =",
        np.mean(
            valid_unc
        )
        if len(valid_unc) > 0
        else np.nan
    )

    print(
        "Puntos con u_MC <= 10 % =",
        np.sum(
            df_hstar[
                "u_MC_percent"
            ] <= 10.0
        ),
        "/",
        len(
            df_hstar
        )
    )
```

```python

# ============================================================
# ANÁLISIS DE SENSIBILIDAD A LA GEOMETRÍA DEL FANTOMA
# HETEROGÉNEO vs. HOMOGÉNEO
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import SimpleITK as sitk
from pathlib import Path

# ============================================================
# 1. CONFIGURACIÓN
# ============================================================

BASE_DIR = (
    Path.home()
    / "OpenGATE"
    / "pruebas_sensibilidad"
    / "output_sensibilidad"
)

ANALYSIS_DIR = BASE_DIR / "analisis_fantoma"
ANALYSIS_DIR.mkdir(parents=True, exist_ok=True)

# IMPORTANTE:
# Colocar el número REAL de historias de cada corrida.

PHANTOM_RUNS = {
    "Heterogéneo": {
        "folder": "scoring_10cm",
        "n_primaries": 10_000_000,
        "n_norm": 10_000_000,
    },
    "Homogéneo": {
        "folder": "fantoma_homogeneo",
        "n_primaries": 10_000_000,
        "n_norm": 10_000_000,
    },
}

SCORING_SIZE_CM = 10.0

# Se reutiliza el kerma de referencia definido en
# el análisis de sensibilidad anterior.
# Ejecutar antes la celda que define esta variable.
KA_REFERENCE_MGY = float(KA_R_SPEKPY_MGY)

# ============================================================
# 2. CARGAR RESULTADOS DE LAS DOS SIMULACIONES
# ============================================================

rows = []
normalizations = []

for model, config in PHANTOM_RUNS.items():

    folder = BASE_DIR / config["folder"]

    main_dir = folder / "main"
    norm_dir = folder / "normalization"

    n_main = config["n_primaries"]
    n_norm = config["n_norm"]

    if n_main <= 1 or n_norm <= 0:
        raise ValueError(
            f"Número de historias inválido: {model}"
        )

    # --------------------------------------------------------
    # NORMALIZACIÓN DEL IRP
    # --------------------------------------------------------

    irp_file = norm_dir / "irp_air_dose.mhd"

    if not irp_file.exists():
        raise FileNotFoundError(irp_file)

    irp_image = sitk.ReadImage(str(irp_file))

    irp_raw_gy = float(
        np.squeeze(
            sitk.GetArrayFromImage(irp_image)
        )
    )

    if not np.isfinite(irp_raw_gy) or irp_raw_gy <= 0:
        raise ValueError(
            f"Resultado IRP inválido: {model}"
        )

    irp_per_primary = irp_raw_gy / n_norm

    equivalent_primaries = (
        KA_REFERENCE_MGY * 1e-3
        / irp_per_primary
    )

    normalizations.append({
        "Modelo": model,
        "IRP_Gy": irp_raw_gy,
        "IRP_Gy_por_primario": irp_per_primary,
        "Primarios_equivalentes": equivalent_primaries,
    })

    # --------------------------------------------------------
    # CALCULAR H*(10) EN LOS DIEZ PUNTOS
    # --------------------------------------------------------

    for point_name, point in MEASUREMENT_POINTS.items():

        root_file = (
            main_dir
            / f"{point_name}_hstar_phase_space.root"
        )

        if not root_file.exists():
            raise FileNotFoundError(root_file)

        result = calculate_hstar10_point(
            root_path=root_file,
            n_primaries=n_main,
            equivalent_real_primaries=equivalent_primaries,
            exposure_time_s=EXPOSURE_TIME_S,
            scoring_size_cm=SCORING_SIZE_CM,
        )

        mc = float(result["Hstar_rate_mSv_h"])
        u_mc_percent = float(result["u_MC_percent"])

        exp = float(point["measured_msv_h"])
        u_exp = float(point["u_measured_msv_h"])

        u_mc = abs(mc) * u_mc_percent / 100.0

        u_combined = np.hypot(u_mc, u_exp)

        rows.append({
            "Modelo": model,
            "Punto": point_name,
            "MC_mSv_h": mc,
            "u_MC_mSv_h": u_mc,
            "u_MC_percent": u_mc_percent,
            "Experimental_mSv_h": exp,
            "u_exp_mSv_h": u_exp,
            "Diferencia_exp_percent":
                100.0 * (mc - exp) / exp,
            "z_exp":
                (mc - exp) / u_combined
                if u_combined > 0 else np.nan,
        })

df_phantom = pd.DataFrame(rows)
df_norm_phantom = pd.DataFrame(normalizations)

# ============================================================
# 3. COMPARACIÓN DIRECTA ENTRE FANTOMAS
# ============================================================

het = (
    df_phantom[
        df_phantom["Modelo"] == "Heterogéneo"
    ]
    .set_index("Punto")
)

hom = (
    df_phantom[
        df_phantom["Modelo"] == "Homogéneo"
    ]
    .set_index("Punto")
)

order = list(MEASUREMENT_POINTS)

het = het.loc[order]
hom = hom.loc[order]

df_compare = pd.DataFrame({
    "Punto": order,
    "Heterogeneo_mSv_h":
        het["MC_mSv_h"].to_numpy(),
    "Homogeneo_mSv_h":
        hom["MC_mSv_h"].to_numpy(),
    "u_heterogeneo_mSv_h":
        het["u_MC_mSv_h"].to_numpy(),
    "u_homogeneo_mSv_h":
        hom["u_MC_mSv_h"].to_numpy(),
    "Experimental_mSv_h":
        het["Experimental_mSv_h"].to_numpy(),
})

a = df_compare["Heterogeneo_mSv_h"].to_numpy()
b = df_compare["Homogeneo_mSv_h"].to_numpy()

ua = df_compare["u_heterogeneo_mSv_h"].to_numpy()
ub = df_compare["u_homogeneo_mSv_h"].to_numpy()

# Cambio relativo respecto del modelo heterogéneo
df_compare["Cambio_percent"] = (
    100.0 * (b - a) / a
)

# Diferencia entre simulaciones,
# suponiendo corridas estadísticamente independientes
u_difference = np.sqrt(ua**2 + ub**2)

df_compare["z_geometria"] = np.divide(
    b - a,
    u_difference,
    out=np.full(len(a), np.nan),
    where=u_difference > 0
)

# ============================================================
# 4. TABLAS DE RESULTADOS
# ============================================================

print("\nNORMALIZACIÓN:")
display(df_norm_phantom)

print("\nCOMPARACIÓN ENTRE FANTOMAS:")
display(df_compare.round(4))

print("\nRESUMEN:")

print(
    "Mediana de |cambio relativo|:",
    f"{df_compare['Cambio_percent'].abs().median():.2f} %"
)

print(
    "Cambio relativo absoluto máximo:",
    f"{df_compare['Cambio_percent'].abs().max():.2f} %"
)

print(
    "Puntos con |z_geometria| > 2:",
    int((df_compare["z_geometria"].abs() > 2).sum()),
    "de",
    len(df_compare)
)

for model, group in df_phantom.groupby("Modelo"):

    print(
        f"\n{model}:"
    )

    print(
        "Mediana incertidumbre MC:",
        f"{group['u_MC_percent'].median():.2f} %"
    )

    print(
        "Mediana diferencia experimental absoluta:",
        f"{group['Diferencia_exp_percent'].abs().median():.2f} %"
    )

# ============================================================
# 5. EXPORTAR TABLAS
# ============================================================

df_phantom.to_csv(
    ANALYSIS_DIR / "resultados_fantomas.csv",
    index=False
)

df_compare.to_csv(
    ANALYSIS_DIR / "comparacion_fantomas.csv",
    index=False
)

df_norm_phantom.to_csv(
    ANALYSIS_DIR / "normalizacion_fantomas.csv",
    index=False
)

# ============================================================
# 6. GRÁFICO 1: TASAS DE DOSIS
# ============================================================

x = np.arange(len(order))

fig, ax = plt.subplots(figsize=(12, 5.5))

ax.errorbar(
    x - 0.15,
    a,
    yerr=ua,
    fmt="o",
    capsize=3,
    label="Fantoma heterogéneo"
)

ax.errorbar(
    x + 0.15,
    b,
    yerr=ub,
    fmt="s",
    capsize=3,
    label="Fantoma homogéneo"
)

ax.errorbar(
    x,
    df_compare["Experimental_mSv_h"],
    fmt="^",
    color="black",
    markersize=5,
    label="Experimental"
)

ax.set_yscale("log")

ax.set_xticks(x)
ax.set_xticklabels(order, rotation=45, ha="right")

ax.set_ylabel(r"Tasa $\dot{H}^*(10)$ (mSv/h)")
ax.set_xlabel("Posición experimental")

ax.legend()
ax.grid(axis="y", alpha=0.3)

plt.tight_layout()

plt.savefig(
    ANALYSIS_DIR / "01_tasas_fantomas.pdf",
    bbox_inches="tight"
)

plt.savefig(
    ANALYSIS_DIR / "01_tasas_fantomas.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 7. GRÁFICO 2: CAMBIO RELATIVO ENTRE FANTOMAS
# ============================================================

fig, ax = plt.subplots(figsize=(12, 5))

ax.bar(
    x,
    df_compare["Cambio_percent"],
    edgecolor="black"
)

ax.axhline(0, color="black", linewidth=1)

ax.set_xticks(x)
ax.set_xticklabels(order, rotation=45, ha="right")

ax.set_ylabel(
    "Cambio relativo homogéneo vs. heterogéneo (%)"
)

ax.set_xlabel("Posición experimental")

ax.grid(axis="y", alpha=0.3)

plt.tight_layout()

plt.savefig(
    ANALYSIS_DIR / "02_cambio_relativo_fantomas.pdf",
    bbox_inches="tight"
)

plt.savefig(
    ANALYSIS_DIR / "02_cambio_relativo_fantomas.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 8. GRÁFICO 3: DIFERENCIA NORMALIZADA ENTRE FANTOMAS
# ============================================================

fig, ax = plt.subplots(figsize=(12, 5))

ax.bar(
    x,
    df_compare["z_geometria"],
    edgecolor="black"
)

ax.axhline(2, linestyle="--", label="Referencia +2")
ax.axhline(-2, linestyle="--", label="Referencia -2")
ax.axhline(0, color="black", linewidth=1)

ax.set_xticks(x)
ax.set_xticklabels(order, rotation=45, ha="right")

ax.set_ylabel(
    "Diferencia normalizada entre simulaciones"
)

ax.set_xlabel("Posición experimental")

ax.legend()
ax.grid(axis="y", alpha=0.3)

plt.tight_layout()

plt.savefig(
    ANALYSIS_DIR / "03_diferencia_normalizada_fantomas.pdf",
    bbox_inches="tight"
)

plt.savefig(
    ANALYSIS_DIR / "03_diferencia_normalizada_fantomas.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("\nAnálisis completado.")
print("Resultados guardados en:", ANALYSIS_DIR)



# Normalización común para ambas geometrías
IRP_PER_PRIMARY_REFERENCE = (
    df_norm_phantom.loc[
        df_norm_phantom["Modelo"] == "Heterogéneo",
        "IRP_Gy_por_primario"
    ].iloc[0]
)

N_EQ_COMMON = (
    KA_REFERENCE_MGY * 1e-3
    / IRP_PER_PRIMARY_REFERENCE
)

print("Primarios equivalentes comunes:", N_EQ_COMMON)
```

```python

# ============================================================
# APLICAR NORMALIZACIÓN COMÚN A AMBOS FANTOMAS
# ============================================================

# Recuperar las normalizaciones originales
n_eq_het = df_norm_phantom.loc[
    df_norm_phantom["Modelo"] == "Heterogéneo",
    "Primarios_equivalentes"
].iloc[0]

n_eq_hom = df_norm_phantom.loc[
    df_norm_phantom["Modelo"] == "Homogéneo",
    "Primarios_equivalentes"
].iloc[0]

# Factor de corrección para el modelo homogéneo
factor_hom = N_EQ_COMMON / n_eq_hom

# Crear una copia para no modificar los resultados originales
df_phantom_common = df_phantom.copy()

mask_hom = df_phantom_common["Modelo"] == "Homogéneo"

# Corregir tasas e incertidumbres absolutas
df_phantom_common.loc[mask_hom, "MC_mSv_h"] *= factor_hom

df_phantom_common.loc[mask_hom, "u_MC_mSv_h"] *= factor_hom

# Recalcular diferencias experimentales
df_phantom_common["Diferencia_exp_percent"] = (
    100.0
    * (
        df_phantom_common["MC_mSv_h"]
        - df_phantom_common["Experimental_mSv_h"]
    )
    / df_phantom_common["Experimental_mSv_h"]
)

# Recalcular diferencias normalizadas
u_total = np.sqrt(
    df_phantom_common["u_MC_mSv_h"]**2
    + df_phantom_common["u_exp_mSv_h"]**2
)

df_phantom_common["z_exp"] = (
    (
        df_phantom_common["MC_mSv_h"]
        - df_phantom_common["Experimental_mSv_h"]
    )
    / u_total
)

print(f"Factor aplicado al homogéneo: {factor_hom:.6f}")

display(
    df_phantom_common[
        [
            "Modelo",
            "Punto",
            "MC_mSv_h",
            "Diferencia_exp_percent",
            "z_exp",
        ]
    ].round(4)
)
```
